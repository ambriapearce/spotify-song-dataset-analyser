```python
import ipywidgets as widgets
from IPython.display import display
import Genre 
from Artist import Artist
import Top5
```

***

```python
# GUI menu 

def widget():
    tab1 = widgets.Output()
    tab2 = widgets.Output()
    tab3 = widgets.Output()
    tabs = widgets.Tab(children=[tab1, tab2, tab3])

    with tab1:
        print("----Genre----")
        Genre.genre()
    with tab2:
        art = Artist("CWDatabase.db")
        print("----Artist----")
        name = input("Please enter an artist's name: ")
        art.artist(name)
        art.close_db()
    with tab3:
        print("----Top 5 Artist----")
        Top5.top5()


    tabs.set_title(0, 'Genre')
    tabs.set_title(1, 'Artist')
    tabs.set_title(2, 'Top 5 Artists')
    
    return tabs



def menu():
    while True:
        print("----MENU----")
        print("[1] Start analysis")
        print("[0] Finish analysis")
        
        option = int(input("Please choose 0 or 1: "))
        if option == 1:
            tab = widget()
            display(tab)
        elif option == 0:
            print("Thank you trying us out!")
            break
        else:
            print("Oh no! Please choose 0 or 1.")
menu()
```
